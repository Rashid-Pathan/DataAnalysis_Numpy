# How to Analyze DataFrames?

# .head()
it shows the first N rows in the data (by default, N=5)


```python
data.head()
```

# .shape
Check the number of rows and columns of the dataframe

```python
data.shape
```

# .index
This attribute provides the index of dataframe

```python
data.index
```

# .columns 
It show the name of each column

```python
data.columns
```

# .dtypes
It shows the data type of each column

```python
data.dtypes
```

# .unique()
In a column it shows all the unique values. It can be applied on single column only not on the whole datafram

```python
data['Weather'].unique()
```

# .nunique()
It show the total nummber of unique values in each column. It can be applied on single column as well as whole dataframe.

```python
data.nunique()
```
